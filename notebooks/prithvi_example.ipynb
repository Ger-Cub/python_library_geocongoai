# Exemple: PrithviClient (geocongoai)

Ce notebook montre un exemple minimal d'utilisation de `PrithviClient`.

Remarque: installez `torch` et `terratorch` si vous voulez exécuter l'exemple.

```python
from geocongoai.ia.prithvi import PrithviClient

client = PrithviClient(model_name="prithvi_eo_v2_300", pretrained=True)
print('Attempting to load model...')
try:
    client.load_model()
    print('Model loaded')
except Exception as e:
    print('Load failed:', e)
```

```python
# Remplacez par un chemin local vers un TIF exemple
tif_path = 'path/to/sample.tif'
try:
    res = client.extract_deep_features(tif_path)
    print('Features keys:', res.keys())
except Exception as e:
    print('Inference failed:', e)
```
